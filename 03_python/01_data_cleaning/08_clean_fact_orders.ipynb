### PostgreSQL Connection

```python
import pandas as pd
from sqlalchemy import create_engine, Numeric

# Use the same DB_URL as your existing MedPlus PostgreSQL setup.
# Recommended: keep the password in an environment variable.
DB_URL = "postgresql://postgres:Sagar29092000@localhost:5432/SwasthyaDB"
engine = create_engine(DB_URL)

print("Connected to PostgreSQL database: SwasthyaDB")
```

### Load the datasets

```python
orders = pd.read_sql_table(
    "fact_orders_raw", con=engine, schema="dbo")
display(orders)
```

### Clean the dataset

```python
orders_clean = orders.copy()
```

```python
orders_clean.info()
```

```python
orders_clean["order_datetime"] = pd.to_datetime(orders_clean["order_datetime"], errors="coerce")
```

```python
orders_clean["promised_delivery_datetime"] = pd.to_datetime(orders_clean["promised_delivery_datetime"], errors="coerce")
```

### Display the clean dataset

```python
display(orders_clean)
```

```python
orders_clean.info()
```

### Load the clean data to PostgreSQL DB

```python
orders_clean.to_sql(
    name="fact_orders",
    con=engine,
    schema="clean",
    if_exists="replace",
    index=False, dtype={'order_value': Numeric(10, 2)}
)

print("Loaded: clean.fact_orders")
```
