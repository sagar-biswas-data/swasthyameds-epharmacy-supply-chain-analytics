### PostgreSQL Connection

```python
import pandas as pd
from sqlalchemy import create_engine, Numeric

# Use the same DB_URL as your existing MedPlus PostgreSQL setup.
# Recommended: keep the password in an environment variable.
DB_URL = "postgresql://postgres:Sagar29092000@localhost:5432/SwasthyaDB"
engine = create_engine(DB_URL)

print("Connected to PostgreSQL database: SwasthyaDB")
```

### Load the datasets

```python
deliveries = pd.read_sql_table(
    "fact_deliveries_raw", con=engine, schema="dbo")
display(deliveries)
```

### Clean the dataset

```python
deliveries_clean = deliveries.copy()
```

```python
deliveries_clean.info()
```

```python
for column in [
    'order_received_datetime', 'picking_start_datetime', 'picking_completed_datetime', 'packing_start_datetime',
    'packing_completed_datetime', 'de_assigned_datetime', 'de_arrival_datetime', 'order_picked_datetime', 'out_for_delivery_datetime',
    'customer_arrival_datetime', 'delivered_datetime']:
    deliveries_clean[column] = pd.to_datetime(deliveries_clean[column], errors="coerce")
```

### Display the clean dataset

```python
deliveries_clean.info()
```

### Load the clean data to PostgreSQL DB

```python
deliveries_clean.to_sql(
    name="fact_deliveries",
    con=engine,
    schema="clean",
    if_exists="replace",
    index=False, dtype={'delivery_distance_km': Numeric(10, 2)}
)

print("Loaded: clean.fact_deliveries")
```
